# QuickSort Algorithm

## Algorithm Overview
- QuickSort is a **divide-and-conquer** algorithm:
  1. Choose a **pivot** (first, middle, or random element).
  2. **Partition** the array:
     - Elements smaller than the pivot go to the left.
     - Elements greater than the pivot go to the right.
  3. Recursively apply QuickSort to the left and right subarrays.

---

## Steps

### 1. Partition Function:
- Initialize pointers: `i = low + 1`, `j = high`.
- Move `i` right until `A[i] >= pivot`.
- Move `j` left until `A[j] < pivot`.
- Swap `A[i]` and `A[j]` if `i < j`.
- Finally, swap the pivot with `A[j]`.
- Return `j` (pivot's final position).

### 2. QuickSort Function:
- **Base Case:** Stop recursion if `low >= high`.
- Partition the array and get the pivot index.
- Recursively sort:
  - Left subarray: `quicksort(A, low, pivot_index - 1)`
  - Right subarray: `quicksort(A, pivot_index + 1, high)`

---

## Pseudocode
```plaintext
Function QuickSort(A, low, high):
    If low < high:
        pivot_index = Partition(A, low, high)
        QuickSort(A, low, pivot_index - 1)  # Left
        QuickSort(A, pivot_index + 1, high)  # Right

Function Partition(A, low, high):
    pivot = A[low]
    i, j = low + 1, high
    While i <= j:
        While i <= j and A[i] <= pivot:
            i++
        While i <= j and A[j] > pivot:
            j--
        If i < j:
            Swap(A[i], A[j])
    Swap(A[low], A[j])
    Return 
```


## Key Points
### Pivot Selection
The pivot determines partition efficiency:
- *First Element:* Simple but prone to worst-case performance on sorted or reverse-sorted arrays.
- *Random Element:* Helps avoid worst-case scenarios.
- *Median-of-Three:* Median of the first, middle, and last elements ensures balanced partitions.




#### Time Complexity
- *Best/Average Case*: O(𝑛log⁡𝑛)
- *Worst Case*: O(𝑛2) when pivot selection is poor (e.g., always the smallest or largest element).
- *Why O(nlogn):* Recursion divides the array into approximately equal parts, and partitioning takes O(n) at each level.

#### Space Complexity
- O(logn) for the recursion stack in the best case.
- O(n) in the worst case due to deep recursion for poor pivot choices.

````python
# QuickSort Implementation

def partition(A, low, high):
    """
    Partition the array around a pivot such that:
    - Elements smaller than the pivot are moved to the left.
    - Elements greater than the pivot are moved to the right.
    Returns the index of the pivot after partitioning.
    """
    pivot = A[low]  # Choose the first element as the pivot
    i, j = low + 1, high

    # Partition process
    while i <= j:
        # Move 'i' to the right until an element >= pivot is found
        while i <= j and A[i] <= pivot:
            i += 1
        
        # Move 'j' to the left until an element < pivot is found
        while i <= j and A[j] > pivot:
            j -= 1
        
        # Swap elements if the pointers haven't crossed
        if i < j:
            A[i], A[j] = A[j], A[i]

    # Place the pivot in its correct position
    A[low], A[j] = A[j], A[low]

    return j  # Return the index of the pivot


def quicksort(A, low, high):
    """
    Recursively apply QuickSort to sort the array.
    A: The array to be sorted.
    low: Starting index of the subarray.
    high: Ending index of the subarray.
    """
    if low < high:
        # Partition the array and get the pivot index
        pivot_index = partition(A, low, high)
        
        # Recursively sort the left and right partitions
        quicksort(A, low, pivot_index - 1)
        quicksort(A, pivot_index + 1, high)



# Input array
A = [10, 16, 8, 12, 15, 6, 3, 9, 5]

# Print the initial array
print("Initial Array:", A)

# Call QuickSort
quicksort(A, 0, len(A) - 1)

# Print the sorted array
print("Sorted Array:", A)
````
