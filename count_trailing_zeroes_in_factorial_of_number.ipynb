# Counting number of trailing zeroes in a factorial of a number

**Brute force or native approach**
1. Find factorial of a number
2. Once you get the factorial of a number, iteratively divide it by 10 and count how many times it is divisible by 10 and return the count. 
This has a time complexity of *O(n)* and space complexity of *O(1)*

Example - 
n = 10
factorial of 10 is 3628800.
3628800 is divisible by 10 twice. Hence, the trailing zeroes will be two.

*pseuod code*

```
zero_count = 0
while(value divisible by 10):
    value = value/10
    zero_count +=1

```


````python
# Let's see the code


def factorial(n):
    sum = 1
    # looping from 1 to n
    for i in range(1, n+1):
        sum = sum * i
    return sum

def count_trailing_zeroes(input):
    sum = factorial(input)
    print("sum is ", sum)
    count = 0
    while(sum%10 == 0):
        # integer division
        sum = sum//10
        count+=1
    return count
````

**Optimized Approach**

Instead of checking if the factorial value is divisible by 10, we can find the count of prime factors( 2, 5) because if a value is divisible by 10 it essentialy means it is also divisible by both 2 and 5. <br>
And even so, we dont need to count number of 2's because any value who has 2 and 5 as prime factors will generally have higher count of 2factors than 5. <br>
Let's say fact(5). It is 
5 x 4 x 3 x 2 x 1. This is 5 x 2^3 x 3. This holds  true for all the others. <br> 
Lets count the total number of 5s in prime factors of n!


Let's find a simple way to count. <br>
Consider fact(n). <br>
Calculating floor(n/5) will give us number of 5's in prime factors of n!. <br>
Fact(5) has one 5 as it is essentially 5 x 4 x 3 x 2 x 1.<br>
fact(10) = 10 x 9 x 8 x 7 x 6 x 5 x 4 x 3 x 2 x 1 = 5^2 x .... ==> 2 5's <br>
fact(15) ==> 3 5's <br>
fact(20) ==> 4 5's <br>

However, some numbers like 25, 125, etc., contribute more than one 5 because they are powers of 5. For example:<br>
In 28!, the number 25 contributes an additional 5, resulting in a total of six 5s.

*In the code implementation, we optimize by just finding the number of times n is divisible by 5 and add it to the count as long as the n is greater or than 5.*

````python
def count_trail_zeroes(n):
    count = 0
    while(n>=5):
        # integer division
        n = n//5
        count = count+n
    return count    
````

How Extra 5s Are Counted:<br>
First Iteration:<br>
n is divided by 5, counting all multiples of 5 (single 5s).
For example, if n = 28, the result of n // 5 is 5 (counting numbers 5, 10, 15, 20, and 25).
<br>

Subsequent Iterations:<br>
In the next iteration, n is further divided by 5. This counts multiples of 25 (which contribute an extra 5 in their factorization).
Continuing this process automatically handles higher powers of 5 like 125, 625, etc., if they exist.

Repeated division by 5 <br>
Time Complexity = O(logn base 5)<br>
Space Complexity = O(1)
